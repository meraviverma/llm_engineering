# STEP 1: Transcribe Audio

## Option 1: Use Open Source for Transcription - Hugging Face Pipelines

```python
from transformers import pipeline

pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium.en",
    dtype=torch.float16,
    device='cuda',
    return_timestamps=True
)

result = pipe(audio_filename)
transcription = result["text"]
print(transcription)
```

```python
open_source_transcription = transcription
```

## Option 2: Use OpenAI for Transcription

```python
# Sign in to OpenAI using Secrets in Colab

AUDIO_MODEL = "gpt-4o-mini-transcribe"

openai_api_key = userdata.get('OPENAI_API_KEY')
openai = OpenAI(api_key=openai_api_key)
transcription = openai.audio.transcriptions.create(model=AUDIO_MODEL, file=audio_file, response_format="text")
print(transcription)
```

```python
display(Markdown(open_source_transcription))
print("\n\n")
display(Markdown(transcription))
```

# STEP 2: Analyze & Report

```python
system_message = """
You produce minutes of meetings from transcripts, with summary, key discussion points,
takeaways and action items with owners, in markdown format without code blocks.
"""

user_prompt = f"""
Below is an extract transcript of a Denver council meeting.
Please write minutes in markdown without code blocks, including:
- a summary with attendees, location and date
- discussion points
- takeaways
- action items with owners

Transcription:
{transcription}
"""

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
  ]
```

```python
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4"
)
```

```python
tokenizer = AutoTokenizer.from_pretrained(LLAMA)
tokenizer.pad_token = tokenizer.eos_token
inputs = tokenizer.apply_chat_template(messages, return_tensors="pt").to("cuda")
streamer = TextStreamer(tokenizer)
model = AutoModelForCausalLM.from_pretrained(LLAMA, device_map="auto", quantization_config=quant_config)
outputs = model.generate(inputs, max_new_tokens=2000, streamer=streamer)
```

```python
response = tokenizer.decode(outputs[0])
```

```python
display(Markdown(response))
```

## Nootebook Explanation

This notebook demonstrates a pipeline for generating meeting minutes from an audio file. The process involves two main steps:

1.  **Audio Transcription**: Converting the spoken audio into text.
2.  **Meeting Minutes Generation**: Using a Large Language Model (LLM) to process the transcript and generate structured meeting minutes, including a summary, discussion points, takeaways, and action items.

### Methods and Functions Used:

#### 1. Audio Transcription (Hugging Face Transformers Pipeline)

This notebook primarily utilizes the `transformers` library from Hugging Face for audio transcription. Specifically, it employs a `pipeline` for `automatic-speech-recognition` with a pre-trained Whisper model.

*   **`transformers.pipeline`**: A high-level API provided by the Hugging Face library that abstracts away much of the complexity of using pre-trained models. For automatic speech recognition, it handles tasks like audio loading, preprocessing, model inference, and text decoding.
    *   **`model="openai/whisper-medium.en"`**: Specifies the use of the `Whisper-medium.en` model, an OpenAI-developed model fine-tuned for English speech recognition. Whisper models are known for their robustness and accuracy across various audio conditions.
    *   **`dtype=torch.float16`**: Configures the model to use 16-bit floating-point precision (half-precision). This can significantly reduce memory usage and speed up computation on compatible hardware (like GPUs) without a substantial loss in accuracy.
    *   **`device='cuda'`**: Directs the pipeline to run the model on a CUDA-enabled GPU. This is crucial for performance with large models and large audio inputs.
    *   **`return_timestamps=True`**: An option to get timestamps for each word or segment, which can be useful for more detailed analysis or synchronization.

*   **`pipe(audio_filename)`**: Invokes the configured pipeline to transcribe the audio file specified by `audio_filename`.
*   **`result["text"]`**: Extracts the full transcribed text from the pipeline's output.
*   **`open_source_transcription`**: A variable used to store the output from the Whisper transcription, making it available for subsequent steps.

#### 2. Meeting Minutes Generation (Google Gemini API)

After transcription, the text is fed into the Google Gemini API to generate structured meeting minutes.

*   **`google.generativeai`**: The Python client library for interacting with Google's Generative AI models, including Gemini.
*   **`userdata.get('GOOGLE_API_KEY')`**: Retrieves the securely stored Google API key from Colab's user data/secrets manager. This is a best practice for handling sensitive credentials.
*   **`genai.configure(api_key=GOOGLE_API_KEY)`**: Initializes the Gemini API client with the obtained API key.
*   **`GEMINI_MODEL_NAME = "gemini-1.5-flash"`**: Defines the specific Gemini model to be used. `gemini-1.5-flash` is a fast and efficient model suitable for many text generation tasks.
*   **`genai.GenerativeModel(GEMINI_MODEL_NAME)`**: Creates an instance of the Gemini generative model.
*   **`system_message` and `user_prompt`**: These strings define the instructions and content provided to the LLM. The `system_message` establishes the persona and desired output format (markdown, summary, discussion points, takeaways, action items). The `user_prompt` provides the actual transcript and reiterates the desired structure.
*   **`full_gemini_prompt = f"{system_message}\n\n{user_prompt}"`**: For Gemini models (especially older versions or specific configurations), it's often effective to concatenate the system and user instructions into a single prompt for `generate_content`.
*   **`gemini_model.generate_content(full_gemini_prompt)`**: Makes the API call to the Gemini model, sending the combined prompt for text generation. The model then processes the prompt and returns its generated response.
*   **`gemini_response.text`**: Extracts the text content from the Gemini model's response.
*   **`IPython.display.Markdown` and `display()`**: Used to render the generated meeting minutes (which are in Markdown format) directly within the Colab notebook for easy readability.
